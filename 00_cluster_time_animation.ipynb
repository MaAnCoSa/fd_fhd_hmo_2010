```python
import pandas as pd
import numpy as np
import geopandas as gpd
import matplotlib.pyplot as plt
import contextily as cx
import imageio
import os
```

```python
data_df = pd.read_csv("gdf_clustered.csv")
gdf = gpd.GeoDataFrame(
    data_df, geometry=gpd.points_from_xy(data_df["x"], data_df["y"]), crs="EPSG:4326"
)
clusters = pd.Categorical(gdf["cluster_kmeans"], categories=set(gdf["cluster_kmeans"]), ordered=False)
gdf["cluster_kmeans"] = clusters
gdf.head()
```

```python
weeks = gdf["week"].unique()
weeks = np.arange(weeks.min(), weeks.max()+1)
weeks
```

```python
folder = "./mapa_clusters_semanal"
weekly_clustered_map_images = []
```

```python
for i, week_num in enumerate(weeks):
    if i == 0:
        week_gdf = gdf[gdf["week"] == weeks[0]]
    else:
        week_gdf = gpd.GeoDataFrame( pd.concat( [week_gdf, gdf[gdf["week"] == weeks[i]]], ignore_index=True) )

    fig, ax = plt.subplots(figsize=(10, 10))
    week_gdf.plot(ax=ax, figsize=(8, 8), alpha=0.5, column="cluster_kmeans", legend=True, cmap='tab10', markersize=8)
    ax.set_title(f"Casos por Cluster - Semana {week_num}")
    ax.set_xlim(-111.1, -110.85)
    ax.set_ylim(28.97, 29.25)

    cx.add_basemap(ax=ax, crs="EPSG:4326", source=cx.providers.CartoDB.Voyager)

    filename = f'mapa_clusters_semanal_{week_num}.png'
    plt.savefig(os.path.join(folder,filename))
    weekly_clustered_map_images.append(imageio.imread(os.path.join(folder,filename)))

imageio.mimsave(os.path.join(folder,'mapa_clusters_semanal.gif'), weekly_clustered_map_images, fps=3)
```

## Animation for Count

```python
gdf.head()
```

```python
cluster_dummies = pd.get_dummies(gdf["cluster_kmeans"], prefix="cluster", dtype="int")
gdf["cluster_0"] = cluster_dummies["cluster_0"]
gdf["cluster_1"] = cluster_dummies["cluster_1"]
gdf["cluster_2"] = cluster_dummies["cluster_2"]
```

```python
# Spike count by day

# Actual counts:
gdf_counts = pd.DataFrame(gdf.groupby("dayofyear")["FD_FHD"].count())
gdf_counts["week"] = gdf.groupby("dayofyear")["week"].mean().astype("int")
gdf_counts["cluster_0"] = gdf.groupby("dayofyear")["cluster_0"].sum()
gdf_counts["cluster_1"] = gdf.groupby("dayofyear")["cluster_1"].sum()
gdf_counts["cluster_2"] = gdf.groupby("dayofyear")["cluster_2"].sum()

# Cumulative total:
gdf_counts["cluster_0_cumsum"] = gdf_counts["cluster_0"].cumsum()
gdf_counts["cluster_1_cumsum"] = gdf_counts["cluster_1"].cumsum()
gdf_counts["cluster_2_cumsum"] = gdf_counts["cluster_2"].cumsum()
# Rolling mean:
gdf_counts["cluster_0_mean"] = gdf_counts["cluster_0"].rolling(7).mean()
gdf_counts["cluster_1_mean"] = gdf_counts["cluster_1"].rolling(7).mean()
gdf_counts["cluster_2_mean"] = gdf_counts["cluster_2"].rolling(7).mean()

gdf_counts.tail()
```
